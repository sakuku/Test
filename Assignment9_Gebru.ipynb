```python
######################
# Assignment: 9
# Author: Saba Gebru
# Date: March 13, 2026
######################

#Import required libraries
import pandas as pd
import plotly.express as px
import io

# Set a default template for all
import plotly.io as pio
pio.templates.default = "plotly_dark" # Other options: "plotly_dark", "seaborn", "ggplot2"

# Set pandas display option for clean output
pd.set_option('display.max_rows', 12)
pd.set_option('display.max_columns', None)

raw_data = """
House_ID,Neighborhood,Sq_Ft,Price,Bedrooms
101,Downtown,1200,450000,2
102,Suburbs,2500,520000,4
103,Uptown,1800,600000,3
104,Downtown,900,350000,1
105,Suburbs,3100,680000,5
106,Uptown,2100,650000,3
107,Suburbs,2200,490000,4
108,Downtown,1500,510000,2
"""

# Load the data directly from string
df = pd.read_csv(io.StringIO(raw_data.strip()))

print("DataFram Setup Complete. Head of Data:")
print(df.head())
```

```python
print("------------------ Data Types ------------------")
df.info()

print ("\n------------------ Statistical Summary ------------------")
print(df.describe())
```

```python
print("\n Generating Line Chart(Total sales volume (sum of prices) in each neighborhood)\n")


# Aggregate: group by Neighborhood and sum the Price column
neighborhood_sales = df.groupby('Neighborhood', as_index=False)['Price'].sum()


fig_bar = px.bar(
    neighborhood_sales,
    x='Neighborhood',
    y='Price',
    color='Neighborhood',       # Color each bar by its neighborhood
    title='Total Sales Volume by Neighborhood',
    labels={'Price': 'Total Sales Volume ($)', 'Neighborhood': 'Neighborhood'},
    text='Price'                # Show exact dollar value on each bar
)


# Best Practice: place text labels above the bars for easy reading
fig_bar.update_traces(textposition='outside')
fig_bar.show()

```

```python
# Relationship: Does a larger house (Square Footage) always mean a higher Price?

print("\n Generating Scatter Plot (Relationship between Square Footage and Price)\n")


fig_scatter = px.scatter(
    df,
    x='Sq_Ft',                  # X-axis: house size
    y='Price',                  # Y-axis: sale price
    color='Neighborhood',       # 3rd Dimension: color codes dots by neighborhood
    size='Bedrooms',            # 4th Dimension: larger dot = more bedrooms (Bonus!)
    hover_name='House_ID',      # Pro-Tip: shows House ID when you hover over a dot!
    title='Square Footage vs. Sale Price by Neighborhood',
    labels={
        'Sq_Ft': 'Square Footage',
        'Price': 'Sale Price ($)',
        'Bedrooms': 'Bedrooms'
    }
)
fig_scatter.show()
```

```python

# Distribution: What is the most common price range for houses in our portfolio?

print("\n Generating Histogram (Distribution of House Sale Prices)\n")


fig_hist = px.histogram(
    df,
    x='Price',
    nbins=4,                             # Divide prices into 4 buckets as req
    title='Distribution of House Sale Prices',
    labels={'Price': 'Sale Price ($)', 'count': 'Number of Houses'},
    color_discrete_sequence=['#3cc09c']  # Pro-Tip: you can force a specific
)


fig_hist.update_layout(
    yaxis_title='Number of Houses',
    bargap=0.05
)


fig_hist.show()


print("------------------------------------------------------------")
print("Dashboard Complete: All 3 charts generated successfully.")
```
