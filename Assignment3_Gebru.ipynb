```python
# ==============================================
# Assignment 3: The Unit Converter Library
# ==============================================

# ----------------------------------------------
# Step 1: The Temperature Tool
# ----------------------------------------------

def to_fahrenheit(celsius):
    """
    Convert a temperature from Celsius to Fahrenheit.
    """
    return (celsius * 1.8) + 32

# ----------------------------------------------
# Step 2: The Distance Tool
# ----------------------------------------------

def to_miles(kilometers):
    """
    Convert a distance from kilometers to miles.
    """
    return kilometers * 0.621371


# -------------------------
# Step 3: Unit Tests
# -------------------------

# Tempreature tests
assert to_fahrenheit(0) == 32, "0°C should equal 32°F"
assert to_fahrenheit(100) == 212, "100°C should equal 212°F"

# Distance tests
assert to_miles(0) == 0, "0 km should equal 0 miles"

# Floating-point test (allow small difference)
assert abs(to_miles(5) - 3.106855) < 0.01, "5 km should be approximately 3.1 miles"


# --------------------------------------
# Step 4: User Interface (UI)
# --------------------------------------
celsius_input = input("Enter temperature in Celsius: ")
celsius_value = float(celsius_input)

fahrenheit_value = to_fahrenheit(celsius_value)

print(f"{celsius_value} degrees Celsius is equal to {fahrenheit_value} degrees Fahrenheit.")

```
