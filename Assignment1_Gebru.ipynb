```python
# Financial Calculator - Calculates gross pay, tax, and net pay for employees

# Get employee information like name, hours worked and rate
name = input("Enter Name: ")
hours_worked = float(input("Enter Hours: "))
hourly_rate = float(input("Enter Rate: "))

# Calculate pay components
gross_pay = hours_worked * hourly_rate
tax_rate = 0.15
tax_amount = gross_pay * tax_rate
net_pay = gross_pay - tax_amount

# Display formatted pay stub
print(f"\nPay Stub for {name}")
print("---")
print(f"Gross Pay: ${gross_pay:,.2f}")
print(f"Tax (15%): ${tax_amount:,.2f}")
print(f"Net Pay: ${net_pay:,.2f}")

#optional Output example
print("\n")
print("\n*** SINGLE-LINE FORMAT ***")
print(f"Pay Stub for {name} --- Gross Pay: ${gross_pay:,.2f} Tax (15%): ${tax_amount:,.2f} Net Pay: ${net_pay:,.2f}")

```
