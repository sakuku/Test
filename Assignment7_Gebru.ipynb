```python

#========================================
# Assignemnt 7: The Sensor Data Analyst
# Author: Saba Gebru
#========================================

import numpy as np

# Raw hourly temperature readings in Celsius
raw_celsius_list = [22.5, 23.0, 22.8, 24.5, 25.1, 24.9, 26.2, 28.0,
                    29.5, 30.2, 31.1, 31.5, 30.8, 29.9, 28.5, 27.0,
                    26.1, 25.5, 24.8, 24.0, 23.5, 23.1, 22.9, 22.6]

# Step 1:  Convert this list into a NumPy array
celsius = np.array(raw_celsius_list)

print("--- 1. list into a NumPy array ---\n")
print(f"Celsius Array:\n{celsius}")
print(f"Shape: {celsius.shape}\n")  # Output: (24,)
```

```python
# Step 2. Vectorized Conversion (Celsius -> Fahrenheit) ---
print("--- 2. Vectorized Conversion ---\n")

# Formula: Fahrenheit = (Celsius * 1.8) + 32
fahrenheit = (celsius * 1.8) + 32  # using Hint Formula: Fahrenheit = (Celsius * 1.8) + 32

print(f"Fahrenheit Array:\n{fahrenheit}\n")
```

```python
# Step 3. Statistical Baseline (Aggregation)
print("--- 3. Statistical Baseline ---\n")

avg_temp = np.mean(fahrenheit)
max_temp = np.max(fahrenheit)
std_temp = np.std(fahrenheit)

print(f"Average (Mean): {avg_temp:.2f} F\n")
print(f"Maximum Temp:   {max_temp:.2f} F\n")
print(f"Std Dev:        {std_temp:.2f} F\n")
```

```python
# --- 4. Anomaly Detection (Boolean Filtering) ---
print("--- 4. Anomaly Detection ---\n")

# Safety threshold
threshold = 85.0

# 4.1 Create the mask (True/False)
anomaly_mask = fahrenheit > threshold
print(f"Anomaly Mask (Temp > {threshold}F):\n{anomaly_mask}\n")

# 4.2 Apply the mask to extract only dangerous readings
dangerous_readings = fahrenheit[anomaly_mask]
print(f"Dangerous Readings:\n{dangerous_readings}\n")

```

```python
# Step 5. Output Report Summary ---
print("--- Daily Sensor Report ---\n")

print(f"Average Temp: {avg_temp:.2f} F\n")
print(f"Max Temp: {max_temp:.2f} F\n")
print(f"Standard Deviation: {std_temp:.2f} F\n")

if dangerous_readings.size > 0:
    print("WARNING: Anomalies Detected!\n")
    # Optional: round for prettier display (keeps NumPy array format)
    print("Dangerous Readings:", np.round(dangerous_readings, 2))
else:
    print("OK: No anomalies detected.")
```
