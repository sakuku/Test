```python
# Assignment 2 The Grade Analyzer

#step 1: Input the Data
scores = [85, 92, 45, 78, 90, 55, 68, 72, 88, 61]

#step 2: Initialize the counters for student passed and failed
passing  = 0
failing = 0

#Step 3: using for loops. Processing  by iterate through each score to count pass or fail
for score in scores:
  if score >=60:
     passing += 1
  else:
    failing += 1

#Step 4 Analysis manual calculation
#calculate the class average
total = 0
for score in scores:
    total = total + score

class_average = total/len(scores)

#Step 5: The if logic to determine the letter grade based on the average
if class_average >=90:
    letter_grade = "A"
elif class_average>=80:
    letter_grade = "B"
elif class_average>=70:
    letter_grade = "C"
elif class_average>=60:
    letter_grade = "D"
else:
    letter_grade = "F"

#Step 6 Print the output

print("-----Class Analysis----\n")
print(f"Total Students: {len(scores)}")
print(f"Passing: {passing}")
print(f"Failing: {failing}")
print(f"Class Average: {class_average}")
print(f"Final Letter Grade: {letter_grade}")
```
