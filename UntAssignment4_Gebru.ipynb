```python
students = [
    {"name": "Alice", "score": 88},
    {"name": "Bob", "score": 54},
    {"name": "Charlie", "score": 92},
    {"name": "Diana", "score": 75},
    {"name": "Evan", "score": 48},
    {"name": "Fiona", "score": 81},
    {"name": "George", "score": 60}
]


total_score = 0
highest_score = 0
at_risk_students = []


for student in students:
    score = student["score"]
    total_score += score

    if score >highest_score:
      highest_score =score

    if score < 60:
       at_risk_students.append(student["name"])

clsas_average =total_score /len(students)

print("--------Class Report-----------")
print(f"Class Average: {clsas_average:.2f}")
print(f"Highest Score : {highest_score}")
print(f"Students At Risk: {at_risk_students}")
```
