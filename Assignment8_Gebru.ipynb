```python

#-----------------------
# Assignment 8: Customer Database Cleaner
# Author: Saba Gebru
# Date 03/08/2026
#-------------------------

# Step 1: Writting Data to CSV

import pandas as pd

raw_data = """
Cust_ID,Name,Purchase_Amount,Discount_Rate
101,Sarah,150.50,10%
102,Mike,,15%
103,Jenna,200.00,5%
101,Sarah,150.50,10%
104,Alex,99.99,20%
105,David,,0%
"""
FILENAME = "customer_data.csv"

# Write the raw string data into a CSV file
with open(FILENAME, "w") as f:
    f.write(raw_data.strip())

print(f"Setup Complete: '{FILENAME}' created on filesystem.")
print("-" * 30)
```

```python
# Step 2: Load the file and Inspection

df = pd.read_csv(FILENAME)

print("\n--- 2. Initial Inspection ---\n")
print(df.head())

# Check data types and non-null counts
print(f"\n\n DataFrame Types and Nulls:\n")
df.info()
```

```python
# Step 3 Data Cleaning and Preprocessing

print("\n 3. Data Cleaning and Preprocessing")
print("\n Duplicate Rows Found:")
print(f"\n Total duplicates to remove: {df.duplicated().sum()}")

```

```python
#Step 4 Remove Duplicate

df = df.drop_duplicates()

print("\n 4. DataFrame After Removing Duplicates:\n")
print(df.head())
```

```python
#Step 5  Check for missing values
print("\n 5. Missing Values Count:\n")
print(df.isnull().sum())
```

```python
# Step 6  Calculate the mean of the Purchase_Amount column
purchase_mean = df["Purchase_Amount"].mean()
print(f"\n 6. Average Purchase Amount is: {purchase_mean:.2f}")
```

```python
# Step 7. Fill missing Purchase_Amount values with the mean
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(purchase_mean)

print("\n7. DataFrame After Filling Missing Purchase_Amount Values:\n")
print(df.head())
```

```python
# Step 8:  Clean and convert the Discount_Rate column

#  Clean and convert the Discount_Rate column
df["Discount_Rate"] = df["Discount_Rate"].astype(str).str.replace("%", "", regex=False).astype(float)

print("\n8. DataFrame After Cleaning Discount_Rate:\n")
print(df.head())
```

```python
### STEP 9: Final Inspection and Review

print("\n\n--- 9. Final Cleaned Data Review ---\n")
print("\n Final DataFrame Head:\n")
print(df.head())

# Final data types and non-null counts
print("\n Final DataFrame Info:\n")
df.info()
```
