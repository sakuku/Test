```python
# RUN THIS CELL FIRST to generate your data file
log_data = """
[INFO] System started successfully
[WARNING] High memory usage detected
[ERROR] Database connection failed
[INFO] User admin logged in
[ERROR] File not found: config.json
[WARNING] Disk space low
[ERROR] Network timeout
[INFO] Scheduled maintenance complete
"""
with open("server_log.txt", "w") as f:
    f.write(log_data.strip())
print("server_log.txt created!")

```

```python
#Initalize error counter
error_count = 0

#open the log for reading and writing the report
with open("server_log.txt", "r") as log_file:
    with open("error_report.txt", "w") as report_file:
  #Loop throthrough each line in the log file
       for line in log_file:
    #Check if the line contains an ERROR
         if "[ERROR]" in line:
      #write the error line
            report_file.write(line)
            error_count += 1

# Append the summary line to the report
with open("error_report.text", "a") as report_file:
    report_file.write(f"total Errors Found:{error_count}\n")

#Verification: print error_report.txt
with open("error_report.txt", "r") as report_file:
    print(report_file.read())
```
